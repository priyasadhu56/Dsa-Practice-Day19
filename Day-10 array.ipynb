```python
# Find the last even number
arr = [1, 3, 5, 7, 8, 10, 11]
last_even = None
for i in arr:
    if i% 2 == 0:
        last_even = i
print(last_even)
```

```python
# Find the last odd number
arr = [1, 3, 5, 7, 8, 10, 11]
last_odd = None
for i in arr:
    if i% 2 != 0:
        last_odd = i
print(last_odd)
```

```python
# Swap first and last element
arr = [1, 3, 5, 7, 8, 10, 11]
arr[0], arr[-1] = arr[-1], arr[0]
print(arr)
```

```python
# Swap two given positions
arr = [1, 3, 5, 7, 8, 10, 11]
a = 1
b = 4
arr[a], arr[b] = arr[b], arr[a]

print(arr)
```

```python
# Move all zeroes to the beginning
arr = [1, 0, 5, 0, 8, 10, 11]
zero_count = arr.count(0)
arr = [0] * zero_count + [x for x in arr if x != 0]
print(arr)
```
