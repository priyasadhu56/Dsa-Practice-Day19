```python
# Check whether an array contains a zero
arr = [1, 2, 3, 4, 5, 0]
if 0 in arr:
    print("The array contains a zero.")
else:
    print("The array does not contain a zero.")
```

```python
# Count numbers divisible by 5
arr = [5, 10, 15, 20, 25, 30]
count = sum(1 for num in arr if num % 5 == 0)
print("Count of numbers divisible by 5:", count)
```

```python
# Print numbers divisible by both 3 and 5
arr = [1, 2, 3, 4, 5, 15, 30, 45]
divisible_by_3_and_5 = [num for num in arr if num % 3 == 0 and num % 5 == 0]
print("Numbers divisible by both 3 and 5:", divisible_by_3_and_5)
```

```python
# Find the sum of positive numbers
arr = [-1, 2, -3, 4, -5, 6]
sum_of_positive = sum(num for num in arr if num > 0)
print("Sum of positive numbers:", sum_of_positive)
```

```python
# Find the product of all elements
arr = [1, 2, 3, 4, 5]
product = 1
for num in arr:
    product *= num
print("Product of all elements:", product)
```

```python
# Find the second smallest distinct number
arr = [5, 2, 8, 1, 3, 2, 1]
distinct_numbers = list(set(arr))
distinct_numbers.sort()
if len(distinct_numbers) >= 2:
    second_smallest = distinct_numbers[1]
    print("Second smallest distinct number:", second_smallest)
```
