```python
# Count numbers divisible by 3
arr = [34,67,90,24,56,89,23,12,45,78,99,100]
count = 0
for i in arr:
    if i % 3 == 0:
        count += 1
print("Count of numbers divisible by 3:", count)
```

```python
# Print numbers divisible by both 2 and 3
arr = [34,67,90,24,56,89,23,12,45,78,99,100]
for i in arr:
    if i % 2 == 0 and i % 3 == 0:
     print(i)
```

```python
# Find the difference between largest and smallest
arr = [40, 3,57,90,23]
largest = arr[0]
smallest = arr[0]
for i in arr:
    if i > largest:
        largest = i
    if i < smallest:
        smallest = i
difference = largest - smallest
print("Difference between largest and smallest:", difference)
```

```python
# Check whether a target exists
arr = [5,67,3,6,8,9,2]
target = 4
found = False
for i in arr :
    if i == target:
        found = True
        break
if found:
         print("Target exists in the array.")
else:
        print("Target does not exist in the array.")
```

```python
# Count how many times a target appears
arr = [5,67,3,6,8,9,2,3,3]
target = 3
count = 0
for i in arr: 
    if i == target:
        count += 1
print("Target appears", count, "times in the array.")
```

```python
# Find the second-last element
arr = [5,67,3,6,8,9,2]
if len(arr) >= 2:
    second_last = arr[-2]
    print("Second-last element:", second_last)  
```

```python
# Print elements at even indexes
arr = [5,67,3,6,8,9,2]
for i in range(0, len(arr), 2):
    print("Element at even index", i, ":", arr[i])
```

```python
# Print elements at odd indexes
arr = [5,67,3,6,8,9,2]
for i in range(1, len(arr), 2):
    print("Element at odd index", i, ":", arr[i])
```

```python
# Find the sum of elements at even indexes
arr = [5,67,3,6,8,9,2]
sum_even_index = 0
for i in range(0, len(arr), 2):
    sum_even_index += arr[i]
print("Sum of elements at even indexes:", sum_even_index)
```

```python
# Find the sum of elements at odd indexes
arr = [5,67,3,6,8,9,2]
sum_odd_index = 0
for i in range(1, len(arr), 2):
    sum_odd_index += arr[i]
print("Sum of elements at odd indexes:", sum_odd_index)
```
